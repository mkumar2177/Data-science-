# Smiley Face (Matplotlib)

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6,6))

# Face
face = plt.Circle((0,0),1,color='yellow', ec='black', linewidth=2)
ax.add_patch(face)

 # Eyes
left_eye = plt.Circle((-0.35,0.3),0.08,color='black')
right_eye = plt.Circle((0.35,0.3),0.08,color='black')

ax.add_patch(left_eye)
ax.add_patch(right_eye)

# Smile (Upper Half of Circle
theta = np.linspace(np.pi/6, 5*np.pi/6, 100)

x = 0.5 * np.cos(theta)
y = -0.2 - 0.4 * np.sin(theta)

ax.plot(x, y, color='black', linewidth=4)

ax.set_xlim(-1.2,1.2)
ax.set_ylim(-1.2,1.2)
ax.set_aspect('equal')

plt.title("Smiley Face using Matplotlib", fontsize=15)
plt.grid()
plt.show()
```
