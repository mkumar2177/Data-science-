# Seaborn Visualization Dashboard

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")

plt.subplot(2,2,2)
sns.boxplot(data=tips, x="day", y="total_bill")

plt.subplot(2,2,3)
sns.violinplot(data=tips, x="day", y="tip")

plt.subplot(2,2,4)
sns.histplot(data=tips, x="total_bill", kde=True)

plt.tight_layout()
plt.show()
```

# Matplotlib Dashboard

```python
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon","Tue","Wed","Thu","Fri"]
sales = [25,40,35,50,45]
profit = [10,18,15,22,20]

plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.plot(days,sales,marker='o')
plt.title("Sales")

plt.subplot(2,2,2)
plt.bar(days,profit)
plt.title("Profit")

plt.subplot(2,2,3)
plt.pie(sales,labels=days,autopct='%1.1f%%')
plt.title("Sales Distribution")

plt.subplot(2,2,4)
plt.scatter(sales,profit)
plt.title("Sales vs Profit")

plt.tight_layout()
plt.show()
```

# Smiley Face (Matplotlib)

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6,6))

# Face
face = plt.Circle((0,0),1,color='yellow')
ax.add_patch(face)

# Eyes
left_eye = plt.Circle((-0.35,0.3),0.1,color='black')
right_eye = plt.Circle((0.35,0.3),0.1,color='black')

ax.add_patch(left_eye)
ax.add_patch(right_eye)

# Smile
x = np.linspace(-0.45,0.45,100)
y = -0.25 - 0.4*(x**2)

ax.plot(x,y,color='black',linewidth=4)

plt.xlim(-1.2,1.2)
plt.ylim(-1.2,1.2)

ax.set_aspect('equal')
plt.title("Smiley Face using Matplotlib")
plt.grid()

plt.show()
```
