```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as mp

img = Image.open("download (16).jpg")
img = np.array(img)

h, w, c = img.shape

mid_h = h // 2
mid_w = w // 2

img[0:mid_h, 0:mid_w, 0] = np.clip(
    img[0:mid_h, 0:mid_w, 0] + 100, 0, 255)

img[0:mid_h, mid_w:w, 1] = np.clip(
    img[0:mid_h, mid_w:w, 1] + 100, 0, 255)

img[mid_h:h, 0:mid_w, 2] = np.clip(
    img[mid_h:h, 0:mid_w, 2] + 100, 0, 255)

img[mid_h:h, mid_w:w, 0] = np.clip(
    img[mid_h:h, mid_w:w, 0] + 100, 0, 255)

img[mid_h:h, mid_w:w, 1] = np.clip(
    img[mid_h:h, mid_w:w, 1] + 100, 0, 255)

# Show image
mp.imshow(img)
mp.axis("OFF")
mp.show()


```
