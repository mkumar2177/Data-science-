#print all even  no between 0 to 100

```python
for i in range(0,101):
   if i%2==0:
       print(f"{i} is even")
```

```python
#add all even no between 0 to 100
```

```python
sum=0
for i in range(0,101):
    if i%2==0:
        sum+=i
print(sum)
```

```python
#print all odd no between 0 to 100
for i in range(0,101):
    if i%2!=0:
        print(f"{i} is odd")
```

```python
#print sum of all odd no between 0 to 100
sum=0
for i in range(0,101):
       if i%2!=0:
           sum+=i
print(sum)
     
```

```python
# Grade calculator if marks 80 to 100 = A,79 to 60 =B, 59 to 50 = C, below 50=F
marks = int(input("Enter your marks: "))
if marks >=80 and marks<=100:
    print("Grade: A")
elif marks >=60 and marks<80:
    print("Grade: B")
elif marks>=50 and marks<60:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
#grade calculator using for loop
marks=int(input("Enter your marks : "))
for i in range(marks,marks+1):
   if i>=80 and i<=100:
    print("Grade: A")
   elif i>=60 and i<80:
    print ("Grade: B")
   elif i>=50 and i<60:
    print("Grade: C")
   else:
    print("Grade: F")
```

```python
#find the largest no from the list
num=[20,25,70,80,64,36]
largest=num[0]
for n in num:
     if n > largest:
         largest =n
print (largest)
```
